# Sales

```python
sales = pd.read_csv('data/sales.csv', encoding='cp1251', sep=';')
sales
```

```python
sales.drop(index=[0, 1, 2], inplace=True)
sales
```

```python
sales = sales[sales['Количество'] >= 0]
sales
```

```python
sales = sales[sales['ID товара'] <= 100]
sales
```

```python
sales['ID клиента'] = sales['ID клиента'].astype(int)
sales['ID товара'] = sales['ID товара'].astype(int)
sales['ID компании'] = sales['ID компании'].astype(int)

sales.rename(columns={
    'Дата продажи': 'Date',
    'ID товара': 'Product ID',
    'ID клиента': 'Client ID',
    'Количество': 'Count',
    'Скидка, %': 'Discount, %',
    'ID компании': 'Company ID'
}, inplace=True)
```

```python
sales.to_csv('data/sales.csv', index=False)
```
